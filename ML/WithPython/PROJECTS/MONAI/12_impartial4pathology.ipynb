## 3.2 Add New Images

- select tiff images from that folder as shown below

![image](images/impartial_add_images.png)


## 3.3 Annotate Images
If you removed labels at step 2.0 then there are no labels for any of the images.
Use the tools to draw labels, then click add for each cell.
Instead of doing this manually, we have the labels already saved on the client system, so you can load it using steps below.

In the ImPartial UI do the following:
1. Click on image
2. Go to ROI Manager
    1.  Click More
    2.  Click open
    3.  Go to the folder researcher/ImPartialData/Vectra_WC_2CH_tiff/labels/final
    4.  Select the same imageX.zip
    5.  Click open
3. You can now see the labels
4. Click submit on the ImPartial main console --> this will save the label back to moani label server 

As in images below:
![image](images/impartial_add_label.png) 

Repeat for images `image0` and `image4`.

## 3.4 Train a Model
Once you've loaded labels for two images, you can start training.
You can leave the default parameters as they are and train for 10 epochs. 
Click Train.
Now you can look at the logs of MONAI Label and see training running.

## 3.5 Run Inference

Note: You can run inference while training is still in progress. Once you reach epoch 4 or 5 you can run inference on other images as `image1`.

Now you can select and image and click inference.
You can see MONAI Label running inference on that image.
